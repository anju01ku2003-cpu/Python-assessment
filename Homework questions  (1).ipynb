```python
#**Question 1: ATM Withdrawal**

#Write a Python program for ATM withdrawal.

pin = 1234
balance = 5000

user_pin = int(input("Enter your PIN: "))

if user_pin == pin:
    amount = int(input("Enter withdrawal amount: "))

    if amount % 100 == 0:
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal Successful")
            print("Remaining Balance:", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

```python
#Question 2: Student Result System
#Write a Python program that takes marks of 3 subjects.

m1 = int(input("Enter marks of Subject 1: "))
m2 = int(input("Enter marks of Subject 2: "))
m3 = int(input("Enter marks of Subject 3: "))

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    total = m1 + m2 + m3
    percentage = total / 3

    print("Total Marks:", total)
    print("Percentage:", percentage)

    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

```python
#Q4. Employee Salary and Department
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, details in employees.items():
    salary = details["salary"]
    department = details["department"]

    if salary >= 50000:
        if department == "IT":
            print(name, "Senior IT Employee")
        elif department == "HR":
            print(name, "Senior HR Employee")
        else:
            print(name, "Senior Employee")
    else:
        print(name, "Junior Employee")
```

```python
#Q5. Check Even/Odd Values
data = {
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}

for key, value in data.items():
    if value % 2 == 0:
        if value > 20:
            print(key, "Large Even")
        else:
            print(key, "Even")
    else:
        print(key, "Odd")
```

```python
#Q7. Student Eligibility
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}

for name, details in students.items():
    attendance = details["attendance"]
    marks = details["marks"]

    if attendance >= 75:
        if marks >= 40:
            print(name, "Eligible")
        else:
            print(name, "Failed in Exam")
    else:
        print(name, "Not Eligible")
```

```python
#Q8. Bank Account Classification
accounts = {
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}

for name, balance in accounts.items():
    if balance < 0:
        print(name, "Overdraft")
    elif balance == 0:
        print(name, "No Balance")
    elif balance >= 10000:
        print(name, "Premium Account")
    else:
        print(name, "Regular Account")
```

```python
#Q9. Nested Dictionary – Employee Experience
employees = {
    "Rahul": {"salary": 60000, "experience": 6},
    "Amit": {"salary": 40000, "experience": 2},
    "Priya": {"salary": 70000, "experience": 3},
    "Neha": {"salary": 45000, "experience": 4}
}

for name, details in employees.items():
    salary = details["salary"]
    experience = details["experience"]

    if salary >= 50000:
        if experience >= 5:
            print(name, "Senior")
        else:
            print(name, "Mid-Level")
    else:
        if experience >= 3:
            print(name, "Experienced Junior")
        else:
            print(name, "Fresher/Junior")
```

```python
#Q10. Student Marks and Attendance
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, details in students.items():
    marks = details["marks"]
    attendance = details["attendance"]

    if attendance < 75:
        print(name, "Not Eligible")
    else:
        if marks >= 80:
            print(name, "Distinction")
        elif marks >= 60:
            print(name, "First Division")
        elif marks >= 40:
            print(name, "Pass")
        else:
            print(name, "Fail")
```

```python
#Exercise 12. Find the Largest Number
data = [22, 55, 12, 89, 34, 67]

largest = data[0]

for num in data:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
#Exercise 13. Find the Smallest Number
data = [22, 55, 12, 89, 34, 67]

smallest = data[0]

for num in data:
    if num < smallest:
        smallest = num

print("Smallest number:", smallest)
```
